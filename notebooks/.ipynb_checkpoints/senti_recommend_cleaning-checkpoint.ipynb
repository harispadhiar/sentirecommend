## Section 3 – Metadata Cleaning

| Step | Issue | Action |
|------|-------|--------|
| A1 | Fully-null columns (`android_version`, `size`) | Drop columns |
| A2 | Irrelevant scraping columns | Drop columns |
| A3 | Dates stored as wrong types | Convert to datetime |
| A4 | Rating-sum vs total_ratings artefact | Document; keep `total_ratings` |
| A5 | Pricing `price_usd` nulls | Fill with 0 |
| A6 | `updated_on` nulls (34 rows) | Impute with median |
| A7 | `avg_rating` range validation | Validate |
| A8 | Text field whitespace | Strip |
| A9 | Duplicate `app_id` rows | Remove |
| A10 | Category encoding for ML | Engineer `project_category_id` |
| A11 | Cold-start popularity tier | Engineer `popularity_tier` |

```python
# ── A1. Drop columns that are 100% null ─────────────────────────────────────
# android_version and size: 482/482 nulls — completely uninformative.

cols_all_null = [c for c in meta.columns if meta[c].isnull().all()]
print(f'Fully-null columns to drop: {cols_all_null}')

meta.drop(columns=cols_all_null, inplace=True)
print(f'Shape after A1: {meta.shape}')
```

```python
# ── A2. Drop irrelevant columns ──────────────────────────────────────────────
# icon_url      → image URL, not a model feature
# genre_id      → redundant with project_category
# content_rating → constant 'Everyone' across all rows, zero variance
# scraped_at    → pipeline metadata, not a feature

irrelevant_cols = ['icon_url', 'genre_id', 'content_rating', 'scraped_at']
meta.drop(columns=irrelevant_cols, inplace=True)

print(f'Shape after A2 : {meta.shape}')
print(f'Remaining cols : {meta.columns.tolist()}')
```

```python
# ── A3. Convert date columns to proper datetime ──────────────────────────────

# updated_on: stored as Unix timestamp (seconds since epoch)
print('updated_on raw sample  :', meta['updated_on'].dropna().head(3).tolist())
meta['updated_on'] = pd.to_datetime(meta['updated_on'], unit='s', errors='coerce')
print('updated_on converted   :', meta['updated_on'].dropna().head(3).tolist())

print()

# released_on: stored as 'Jan 10, 2023' string format
print('released_on raw sample :', meta['released_on'].dropna().head(3).tolist())
meta['released_on'] = pd.to_datetime(meta['released_on'], format='%b %d, %Y', errors='coerce')
print('released_on converted  :', meta['released_on'].dropna().head(3).tolist())
print(f'released_on nulls remaining : {meta["released_on"].isnull().sum()}')
```

```python
# ── A4. Document rating-sum artefact & remove helper columns ─────────────────
# Cause: Google Play updates aggregate rating counts asynchronously.
# Effect: sum(1-5 star columns) rarely equals total_ratings exactly.
# Decision: total_ratings is the authoritative count.
#           Individual star columns kept as approximate distributions.

meta['_ratings_sum'] = meta[['ratings_1_star','ratings_2_star',
                               'ratings_3_star','ratings_4_star',
                               'ratings_5_star']].sum(axis=1)
n_mismatch = (meta['_ratings_sum'] != meta['total_ratings']).sum()
mean_diff   = (meta['_ratings_sum'] - meta['total_ratings']).abs().mean()
max_diff    = (meta['_ratings_sum'] - meta['total_ratings']).abs().max()

print(f'Rows with mismatch        : {n_mismatch} / {len(meta)}')
print(f'Mean absolute discrepancy : {mean_diff:.1f}')
print(f'Max  absolute discrepancy : {max_diff}')
print()
print('Decision: Keeping total_ratings as authoritative. Dropping helper columns.')
meta.drop(columns=['_ratings_sum', '_rating_diff'], inplace=True, errors='ignore')
```

```python
# ── A5. Standardise pricing columns ──────────────────────────────────────────
# price_usd is 0.0 for all Free/Freemium apps — correct.
# Fill any NaN with 0 as a defensive precaution.

print('price_usd nulls before:', meta['price_usd'].isnull().sum())
meta['price_usd'] = meta['price_usd'].fillna(0.0)
print('price_usd nulls after :', meta['price_usd'].isnull().sum())
print()
print('Pricing model breakdown:')
print(meta['pricing_model'].value_counts().to_string())
```

```python
# ── A6. Impute missing updated_on with median date ───────────────────────────
# 34 apps have no update timestamp.
# Using the median keeps imputation conservative and avoids
# artificially inflating recency scores in the recommendation engine.

null_before   = meta['updated_on'].isnull().sum()
median_update = meta['updated_on'].dropna().median()
meta['updated_on'] = meta['updated_on'].fillna(median_update)

print(f'updated_on nulls before imputation : {null_before}')
print(f'Imputed with median date            : {median_update.date()}')
print(f'updated_on nulls after imputation   : {meta["updated_on"].isnull().sum()}')
```

```python
# ── A7. Validate avg_rating is within [1, 5] ────────────────────────────────

out_of_range = meta[(meta['avg_rating'] < 1) | (meta['avg_rating'] > 5)]
print(f'Rows with avg_rating outside [1-5] : {len(out_of_range)}')
print(f'avg_rating range : {meta["avg_rating"].min()} - {meta["avg_rating"].max()}')
print('✅ All avg_rating values are valid.')
```

```python
# ── A8. Strip whitespace from text columns ───────────────────────────────────

for col in ['app_name', 'developer', 'app_version', 'url']:
    if col in meta.columns:
        meta[col] = meta[col].astype(str).str.strip()

print('✅ Whitespace stripped from: app_name, developer, app_version, url')
print()
print('Sample app_name values:')
print(meta['app_name'].head(5).tolist())
```

```python
# ── A9. Remove duplicate app_ids ─────────────────────────────────────────────

before = len(meta)
meta.drop_duplicates(subset='app_id', keep='first', inplace=True)
print(f'Duplicate app_id rows removed : {before - len(meta)}')
print(f'Shape after A9               : {meta.shape}')
```

```python
# ── A10. Encode project_category as numeric ──────────────────────────────────
# The Surprise SVD model requires numeric category IDs.

cat_map = {
    'Generative Text / Chatbots' : 1,
    'Image Generation'           : 2,
    'AI Coding Assistants'       : 3,
    'Productivity AI'            : 4,
    'Marketing AI'               : 5
}
meta['project_category_id'] = meta['project_category'].map(cat_map)

print('Category encoding:')
for k, v in cat_map.items():
    count = (meta['project_category'] == k).sum()
    print(f'  {v}  ->  {k}  ({count} apps)')
```

```python
# ── A11. Engineer popularity_tier feature ────────────────────────────────────
# Used by the Knowledge-Based cold-start fallback for ranking tools
# that have zero reviews (per project scope section d).

meta['popularity_tier'] = pd.cut(
    meta['avg_rating'],
    bins   = [0, 2.5, 3.5, 4.0, 4.5, 5.01],
    labels = ['Very Low', 'Low', 'Medium', 'High', 'Very High'],
    right  = False
)

print('popularity_tier distribution:')
print(meta['popularity_tier'].value_counts().sort_index().to_string())
```

```python
# ── Metadata: final state ────────────────────────────────────────────────────
print('=' * 55)
print('  METADATA – AFTER CLEANING')
print('=' * 55)
print(f'Shape: {meta.shape}')
print()
print('Remaining nulls:')
remaining = meta.isnull().sum()
nulls_left = remaining[remaining > 0]
print(nulls_left.to_string() if len(nulls_left) > 0 else '  None ✅')
print()
meta.head(3)
```

## Section 4 – Reviews Cleaning

| Step | Issue | Action |
|------|-------|--------|
| B1 | `review_title` 100% null | Drop column |
| B2 | Null `review_text` | Drop rows |
| B3 | Null `user_name` | Drop rows |
| B4 | Dirty text (escaped `\n`, control chars, extra spaces) | Clean function |
| B5 | Trivially short reviews (<3 chars) | Drop rows |
| B6 | Text length feature | Engineer |
| B7 | `review_date` stored as string | Convert to datetime |
| B8 | Year/month features | Engineer |
| B9 | `reply_text` legitimately sparse (~70% null) | Create binary flag |
| B10 | `app_version_review` nulls (~14%) | Fill with 'Unknown' |
| B11 | `star_rating` range validation | Validate |
| B12 | Duplicate `review_id` rows | Remove |
| B13 | Orphan reviews (no matching app) | Remove |
| B14 | `sort_source` scraping artefact | Drop column |
| B15 | `scraped_at` constant value | Drop column |

```python
# ── B1. Drop review_title (100% null) ────────────────────────────────────────
print(f'review_title null count: {reviews["review_title"].isnull().sum()} / {len(reviews)}')
reviews.drop(columns=['review_title'], inplace=True)
print('✅ review_title dropped.')
```

```python
# ── B2. Remove rows with null review_text ───────────────────────────────────
before = len(reviews)
reviews.dropna(subset=['review_text'], inplace=True)
print(f'Rows removed due to null review_text : {before - len(reviews)}')
```

```python
# ── B3. Remove rows with null user_name ──────────────────────────────────────
before = len(reviews)
reviews.dropna(subset=['user_name'], inplace=True)
print(f'Rows removed due to null user_name : {before - len(reviews)}')
```

```python
# ── B4. Clean review_text ─────────────────────────────────────────────────────

def clean_review_text(text):
    \"\"\"
    Cleans raw review text for NLP / sentiment analysis.

    Steps:
      1. Cast to string  — guards against edge-case NaN / float values
      2. Strip leading / trailing whitespace
      3. Replace escaped newlines (\\\\n) left by scraper with a space
      4. Collapse consecutive spaces / tabs to a single space
      5. Remove non-printable / control characters
    \"\"\"
    text = str(text).strip()
    text = text.replace('\\\\n', ' ').replace('\\n', ' ').replace('\\r', ' ')
    text = re.sub(r'[ \\t]+',  ' ', text)                      # collapse spaces
    text = re.sub(r'[^\\x20-\\x7E\\u00A0-\\uFFFF]', '', text)  # remove control chars
    return text.strip()

# Demonstrate on a sample with escaped \\n characters
sample_dirty = reviews[reviews['review_text'].str.contains(r'\\\\n', na=False)]
if len(sample_dirty) > 0:
    dirty_ex = sample_dirty['review_text'].iloc[0]
    print('BEFORE:', repr(dirty_ex[:120]))
    print('AFTER :', repr(clean_review_text(dirty_ex)[:120]))
    print()

reviews['review_text'] = reviews['review_text'].apply(clean_review_text)

# Drop rows that became empty after cleaning
before = len(reviews)
reviews = reviews[reviews['review_text'].str.len() > 0]
print(f'Empty-after-cleaning rows removed : {before - len(reviews)}')
print('✅ review_text cleaned.')
```

```python
# ── B5. Remove trivially short reviews ───────────────────────────────────────
# Reviews shorter than 3 characters (e.g. '.', 'ok') provide no meaningful
# signal for VADER or Hugging Face sentiment models.

before = len(reviews)
reviews = reviews[reviews['review_text'].str.len() >= 3]
print(f'Reviews with <3 characters removed : {before - len(reviews)}')
print(f'Shape after B5                     : {reviews.shape}')
```

```python
# ── B6. Engineer review_text_length feature ──────────────────────────────────
# Longer reviews carry richer sentiment signal.
# This feature can be used as a confidence weight in the hybrid fusion formula:
#   Final_Score = w1 * CF_score + w2 * (sentiment * log(text_length))

reviews['review_text_length'] = reviews['review_text'].str.len()

print('review_text_length statistics:')
print(reviews['review_text_length'].describe().to_string())
```

```python
# ── B7. Parse review_date to datetime ────────────────────────────────────────
print('review_date raw sample:', reviews['review_date'].head(3).tolist())

reviews['review_date'] = pd.to_datetime(reviews['review_date'], errors='coerce')

print('review_date after parsing:', reviews['review_date'].head(3).tolist())
print(f'Unparseable dates         : {reviews["review_date"].isnull().sum()}')
```

```python
# ── B8. Extract year and month for temporal analysis ─────────────────────────
# Useful for tracking sentiment drift over time
# (e.g. detecting user reaction to a bad app update).

reviews['review_year']  = reviews['review_date'].dt.year
reviews['review_month'] = reviews['review_date'].dt.month

print('Review year distribution:')
print(reviews['review_year'].value_counts().sort_index().to_string())
```

```python
# ── B9. Handle reply columns ─────────────────────────────────────────────────
# ~70% of reviews have no developer reply — expected, not a data error.
# Create a binary flag; keep original reply_text and reply_date columns.

reviews['has_developer_reply'] = reviews['reply_text'].notna().astype(int)

reply_rate = reviews['has_developer_reply'].mean() * 100
print(f'Developer reply rate     : {reply_rate:.1f}%')
print(f'Reviews WITH reply       : {reviews["has_developer_reply"].sum():,}')
print(f'Reviews WITHOUT reply    : {(reviews["has_developer_reply"] == 0).sum():,}')
```

```python
# ── B10. Fill app_version_review nulls ───────────────────────────────────────
# ~14% of reviews lack version info — common when users skip app updates.
# Filling with 'Unknown' preserves the row while marking the gap clearly.

null_ver = reviews['app_version_review'].isnull().sum()
reviews['app_version_review'] = reviews['app_version_review'].fillna('Unknown')
print(f'app_version_review nulls filled : {null_ver}')
```

```python
# ── B11. Validate star_rating range [1–5] ────────────────────────────────────
invalid = reviews[(reviews['star_rating'] < 1) | (reviews['star_rating'] > 5)]
print(f'Invalid star_rating rows : {len(invalid)}')
print('✅ All star_rating values are within [1–5].')
print()
print('Star rating distribution (cleaned):')
print(reviews['star_rating'].value_counts().sort_index().to_string())
```

```python
# ── B12. Remove duplicate review_ids ─────────────────────────────────────────
before = len(reviews)
reviews.drop_duplicates(subset='review_id', keep='first', inplace=True)
print(f'Duplicate review_id rows removed : {before - len(reviews)}')
```

```python
# ── B13. Enforce referential integrity ───────────────────────────────────────
# Every review must map to an app present in the cleaned metadata.

valid_app_ids = set(meta['app_id'])
before = len(reviews)
reviews = reviews[reviews['app_id'].isin(valid_app_ids)]
print(f'Orphan reviews removed (no matching app) : {before - len(reviews)}')
```

```python
# ── B14. Drop sort_source ────────────────────────────────────────────────────
# 'relevant' vs 'newest' describes how the scraper fetched the review.
# This is a scraping artefact with zero predictive value.

print('sort_source value counts:')
print(reviews['sort_source'].value_counts().to_string())
reviews.drop(columns=['sort_source'], inplace=True)
print('✅ sort_source dropped.')
```

```python
# ── B15. Drop scraped_at from reviews ────────────────────────────────────────
# Single constant timestamp for all rows (one scrape run) — zero variance.

print('scraped_at unique values:', reviews['scraped_at'].nunique())
reviews.drop(columns=['scraped_at'], inplace=True)
print('✅ scraped_at dropped.')
```

```python
# ── Reviews: final state ─────────────────────────────────────────────────────
print('=' * 55)
print('  REVIEWS – AFTER CLEANING')
print('=' * 55)
print(f'Shape: {reviews.shape}')
print()
print('Remaining nulls:')
remaining = reviews.isnull().sum()
nulls_left = remaining[remaining > 0]
print(nulls_left.to_string() if len(nulls_left) > 0 else '  None ✅')
print()
reviews.head(3)
```

## Section 5 – Cross-Dataset Integrity Check
Verify the two cleaned datasets are consistent and ready for model building.

```python
# ── 5.1  App-ID alignment ────────────────────────────────────────────────────
meta_ids    = set(meta['app_id'].unique())
review_ids  = set(reviews['app_id'].unique())

apps_no_reviews    = meta_ids - review_ids
orphan_review_apps = review_ids - meta_ids

print(f'Apps in metadata                    : {len(meta_ids)}')
print(f'Apps with at least 1 review         : {len(review_ids)}')
print(f'Apps with ZERO reviews (cold-start) : {len(apps_no_reviews)}')
print(f'Review app_ids not in metadata      : {len(orphan_review_apps)}')
```

```python
# ── 5.2  Reviews-per-app distribution ────────────────────────────────────────
rpa = (reviews.groupby('app_id').size()
              .reset_index(name='n_reviews')
              .merge(meta[['app_id','app_name','project_category','avg_rating']],
                     on='app_id', how='left')
              .sort_values('n_reviews', ascending=False))

print('Reviews-per-app statistics:')
print(rpa['n_reviews'].describe().to_string())
print()
print('Top 10 most-reviewed apps:')
rpa.head(10)[['app_name','project_category','avg_rating','n_reviews']]
```

```python
# ── 5.3  Review coverage by project category ─────────────────────────────────
coverage = (reviews.merge(meta[['app_id','project_category']], on='app_id', how='left')
                   .groupby('project_category')
                   .agg(n_reviews=('review_id','count'),
                        avg_star=('star_rating','mean'))
                   .round(2)
                   .sort_values('n_reviews', ascending=False))

print('Review coverage by category:')
coverage
```

```python
# ── 5.4  Star rating distribution (cleaned) ──────────────────────────────────
star_dist = reviews['star_rating'].value_counts().sort_index()
star_pct  = (star_dist / len(reviews) * 100).round(1)

star_summary = pd.DataFrame({'Count': star_dist, '% of Reviews': star_pct})
print('Overall star rating distribution:')
star_summary
```

## Section 6 – Save Cleaned Outputs

```python
# ── Save cleaned CSVs ────────────────────────────────────────────────────────
META_OUT    = 'cleaned_metadata.csv'
REVIEWS_OUT = 'cleaned_reviews.csv'

meta.to_csv(META_OUT, index=False)
reviews.to_csv(REVIEWS_OUT, index=False)

print('✅ Files saved successfully.')
print(f'   {META_OUT}    ->  {meta.shape[0]:,} rows x {meta.shape[1]} columns')
print(f'   {REVIEWS_OUT} ->  {reviews.shape[0]:,} rows x {reviews.shape[1]} columns')
```

```python
# ── Final cleaning summary ───────────────────────────────────────────────────
summary = pd.DataFrame([
    ['Metadata',  meta_raw.shape[0],     meta_raw.shape[1],    meta.shape[0],     meta.shape[1]],
    ['Reviews',   reviews_raw.shape[0],  reviews_raw.shape[1], reviews.shape[0],  reviews.shape[1]],
], columns=['Dataset','Raw Rows','Raw Cols','Clean Rows','Clean Cols'])

summary['Rows Removed'] = summary['Raw Rows'] - summary['Clean Rows']
summary['Cols Removed'] = summary['Raw Cols'] - summary['Clean Cols']

print('=' * 60)
print('  FINAL CLEANING SUMMARY')
print('=' * 60)
print(summary.to_string(index=False))
print()
print('New features engineered – Metadata:')
print('  project_category_id  ->  numeric label for SVD/CF model')
print('  popularity_tier      ->  cold-start knowledge-based fallback')
print()
print('New features engineered – Reviews:')
print('  review_text_length   ->  sentiment confidence weight')
print('  review_year          ->  temporal drift analysis')
print('  review_month         ->  temporal drift analysis')
print('  has_developer_reply  ->  developer engagement quality signal')
print()
print('Next step: Section – Sentiment Analysis (VADER / Hugging Face)')
```
