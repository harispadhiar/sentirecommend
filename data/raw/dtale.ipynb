```python
import pandas as pd
import dtale

# Create a sample DataFrame or load your own data
df = pd.DataFrame([
    {'a': 1, 'b': 2, 'c': 3},
    {'a': 4, 'b': 5, 'c': 6},
    {'a': 7, 'b': 8, 'c': 9}
])

# Launch the D-Tale interface
d = dtale.show(df)

# The interface will open in your web browser or within your Jupyter notebook environment
# You can access the URL or kill the process through the returned object 'd'
d.open_browser()
d.kill()
```
